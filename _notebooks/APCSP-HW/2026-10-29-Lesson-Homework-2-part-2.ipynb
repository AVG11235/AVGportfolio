````python
%% python

# CODE_RUNNER: Popcorn Hack 1 - wildfire dispatch scanner

alerts = [1, 4, 2, 6, 3, 9, 5]

safe_cutoff = 3
critical_cutoff = 8

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Safe alert skipped:", alert)
        continue

    if alert >= critical_cutoff:
        print("Critical alert found:", alert)
        critical_found = True
        break

    action_count += 1
    total_action_score += alert
    print("Dispatch responding to alert:", alert)

print("Safe alerts skipped:", safe_count)
print("Action count:", action_count)
print("Total action score:", total_action_score)
print("Critical found:", critical_found)
````

Prediction: - Alerts below 3 will be skipped using continue.
- Alerts from 3 through 7 will count as action alerts.
- The loop will stop when it reaches alert 9, using break.
- Alert 9 will not be added to the action count or total because the critical check happens first.

Explanation: for alert in alerts examines each alert in order.

````python
%% python

# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# No else needed: if gift cards are absent, add no points for this signal.
if "gift card" in message:
    risk_score += 4

# No else needed: if the amount is not over 100, add no points for this signal.
if amount_requested > 100:
    risk_score += 3

# No else needed: if the sender is known, add no points for this signal.
if not known_contact:
    risk_score += 2

print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, and check the message carefully.")
else:
    print("No risk signals were detected by this scoring system.")
````

Output Expected: Risk score: 9
