````python
%%python

# CODE_RUNNER: Popcorn Hack 1 Check One SFI Part Record

part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001"
}

valid_categories = ["Auto Racing", "Drag Racing"]
existing_spec_numbers = ["1.1", "2.1"]

def evaluate_create_rule(record):
    has_product_name = record["product_name"] != ""
    valid_category = record["category"] in valid_categories
    has_spec_number = record["spec_number"] != ""
    has_effective_date = record["effective_date"] != ""
    duplicate_spec = record["spec_number"] in existing_spec_numbers

    can_create_record = (
        has_product_name
        and valid_category
        and has_spec_number
        and has_effective_date
        and not duplicate_spec
    )

    print(f"Testing Spec {record['spec_number']} ({record['product_name']}):")
    print("  Duplicate Spec:", duplicate_spec)
    print("  Can Create Record:", can_create_record)
    if can_create_record:
        print("  Decision: ACCEPT RECORD (Create in Database)")
    else:
        print("  Decision: REJECT RECORD (Do Not Create)")

print("--- Case 1: Starter Record (Duplicate Spec 1.1) ---")
evaluate_create_rule(part)

# Valid non-duplicate test case
new_valid_part = {
    "product_name": "Bellhousing Safety Shield",
    "category": "Drag Racing",
    "spec_number": "1.5",
    "effective_date": "Sep. 15, 2026"
}

print("\n--- Case 2: New Valid Record (Spec 1.5) ---")
evaluate_create_rule(new_valid_part)
````

````python
%% python

# CODE_RUNNER: Libraries HW plus hacks

import json
import statistics
from datetime import date

# My basketball season
games = [
    {
        "date": "2026-09-14",
        "opponent": "Lincoln",
        "points": 24,
        "minutes": 28
    },
    {
        "date": "2026-09-18",
        "opponent": "Madison",
        "points": 18,
        "minutes": 22
    },
    {
        "date": "2026-09-22",
        "opponent": "Lincoln",
        "points": 31,
        "minutes": 30
    },
    {
        "date": "2026-09-27",
        "opponent": "Roosevelt",
        "points": 27,
        "minutes": 26
    },
    {
        "date": "2026-10-02",
        "opponent": "Madison",
        "points": 22,
        "minutes": 25
    }
]


def summarize_season(season):
    """
    Summarize a list of basketball games.

    Parameters:
        season: list of dictionaries containing date, opponent,
                points, and minutes.

    Returns:
        A dictionary containing total points, average points,
        best game, and points scored against each opponent.
    """

    points = [game["points"] for game in season]

    points_by_opponent = {}

    for game in season:
        opponent = game["opponent"]

        if opponent not in points_by_opponent:
            points_by_opponent[opponent] = 0

        points_by_opponent[opponent] += game["points"]

    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 1),
        "best_game": max(points),
        "points_by_opponent": points_by_opponent
    }


# Create the season report
report = summarize_season(games)

# My next game is October 15, 2026
next_game = date(2026, 10, 15)

days_until = (next_game - date.today()).days

report["days_until_next_game"] = days_until

# Display the report as JSON
print(json.dumps(report, indent=2))


# requirements.txt
# json, statistics, and datetime are part of Python's
# standard library, so they do not need to be installed.
requirements_txt = """pandas
Flask
"""

print()
print("requirements.txt:")
print(requirements_txt)
````
