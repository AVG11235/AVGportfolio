Toggle HW

```html
<!DOCTYPE html>
<html>
<head>
  <title>My Interactive Toggles</title>

  <style>
    body {
      font-family: Arial, sans-serif;
      padding: 30px;
      background-color: #f4f4f4;
    }

    .toggle-section {
      background: white;
      padding: 25px;
      max-width: 600px;
      margin: auto;
      border-radius: 12px;
    }

    .toggle-row {
      margin: 20px 0;
    }

    #toggleCount {
      font-size: 20px;
      font-weight: bold;
      color: #4b0082;
    }

    #status {
      margin-top: 20px;
      padding: 15px;
      background-color: #eee;
      border-radius: 8px;
    }
  </style>
</head>

<body>

  <div class="toggle-section">
    <h1>My Interactive Toggles</h1>

    <!-- Toggle 1: controls the background color -->
    <div class="toggle-row">
      <label>
        Dark Mode
        <input type="checkbox" class="ocs__toggle" id="darkMode">
      </label>
    </div>

    <!-- Toggle 2: controls whether the message is displayed -->
    <div class="toggle-row">
      <label>
        Show Welcome Message
        <input type="checkbox" class="ocs__toggle" id="welcome">
      </label>
    </div>

    <!-- Toggle 3: controls the size of the text -->
    <div class="toggle-row">
      <label>
        Large Text
        <input type="checkbox" class="ocs__toggle" id="largeText">
      </label>
    </div>

    <h2>Toggles Currently On: <span id="toggleCount">0</span></h2>

    <div id="status">
      <p id="welcomeMessage" style="display: none;">
        Welcome to my interactive page!
      </p>

      <p id="instructions">
        Turn on the toggles above to change the page.
      </p>
    </div>
  </div>

  <script>
    // I learned that .checked lets JavaScript determine whether a checkbox
    // is currently turned on (true) or off (false).
    const darkMode = document.getElementById("darkMode");
    const welcome = document.getElementById("welcome");
    const largeText = document.getElementById("largeText");

    const toggleCount = document.getElementById("toggleCount");
    const welcomeMessage = document.getElementById("welcomeMessage");
    const instructions = document.getElementById("instructions");

    // This function checks every toggle and updates the page.
    function updatePage() {
      let numberOn = 0;

      // Check the Dark Mode toggle.
      if (darkMode.checked) {
        numberOn++;
        document.body.style.backgroundColor = "#222";
        document.body.style.color = "white";
      } else {
        document.body.style.backgroundColor = "#f4f4f4";
        document.body.style.color = "black";
      }

      // Check the Welcome Message toggle.
      if (welcome.checked) {
        numberOn++;
        welcomeMessage.style.display = "block";
      } else {
        welcomeMessage.style.display = "none";
      }

      // Check the Large Text toggle.
      if (largeText.checked) {
        numberOn++;
        instructions.style.fontSize = "24px";
      } else {
        instructions.style.fontSize = "16px";
      }

      // Update the number of toggles that are currently turned on.
      toggleCount.textContent = numberOn;
    }

    // I used one function for all three toggles so the counter stays
    // accurate whenever any toggle is changed.
    darkMode.addEventListener("change", updatePage);
    welcome.addEventListener("change", updatePage);
    largeText.addEventListener("change", updatePage);

    // Run the function once when the page loads.
    updatePage();
  </script>

</body>
</html>
```
