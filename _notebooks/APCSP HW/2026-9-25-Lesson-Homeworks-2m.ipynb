  <!-- Toggle 3 controls the text size -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="largeText" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Large Text</span>
  </label>

  <hr>

  <h3>Toggles Currently On: <span id="toggleCount">0</span></h3>

  <p id="welcomeMessage" hidden>
    Welcome! You turned on the welcome message.
  </p>

  <p id="instructions">
    Turn the toggles on and off to change this page.
  </p>

</div>

<script>

  const darkMode = document.getElementById("darkMode");
  const welcome = document.getElementById("welcome");
  const largeText = document.getElementById("largeText");

  const toggleCount = document.getElementById("toggleCount");
  const welcomeMessage = document.getElementById("welcomeMessage");
  const instructions = document.getElementById("instructions");

  // I thought .checked was interesting because JavaScript can use it
  // to tell whether a toggle is currently on (true) or off (false).
  function updatePage() {

    let numberOn = 0;

    // Check the Dark Mode toggle.
    if (darkMode.checked) {
      numberOn++;
      document.body.style.backgroundColor = "#222";
      document.body.style.color = "white";
    } else {
      document.body.style.backgroundColor = "";
      document.body.style.color = "";
    }

    // Check the Welcome Message toggle.
    if (welcome.checked) {
      numberOn++;
      welcomeMessage.hidden = false;
    } else {
      welcomeMessage.hidden = true;
    }

    // Check the Large Text toggle.
    if (largeText.checked) {
      numberOn++;
      instructions.style.fontSize = "25px";
    } else {
      instructions.style.fontSize = "";
    }

    // This number changes every time a toggle is turned on or off.
    toggleCount.textContent = numberOn;
  }

  // The "change" event runs whenever the user flips a toggle.
  darkMode.addEventListener("change", updatePage);
  welcome.addEventListener("change", updatePage);
  largeText.addEventListener("change", updatePage);

  // Run once when the page first loads.
  updatePage();

</script>
